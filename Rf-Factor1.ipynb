```python
import numpy as np

rf_list = []
x_list = []

while True:
    question = input("Print Rf and X ")
    if question.lower() == "stop":
        break
    try:
        rf, x = question.split()
        rf1 = float(rf)
        x1 = float(x)
        if not (0 < rf1 < 1):
            print("Impossible number ")
            continue
        if not (0 < x1 <= 1):
            print("Impossible number ")
            continue
        rf_list.append(rf1)
        x_list.append(x1)
    except ValueError:
         print("Not enough values to unpack or it's not a numbers")

if len(rf_list) < 2:
    print("One point can't define a line")
    exit()
m_rf = np.array(rf_list)
m_x = np.array(x_list)
m_k = ((1-m_rf)/m_rf)

log_k = np.log10(m_k)
log_x = np.log10(m_x)

slope, intercept = np.polyfit(log_x, log_k, 1)
n = -slope
k0 = 10**(intercept)





```
